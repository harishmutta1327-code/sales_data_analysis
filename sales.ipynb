```python
# Sales Data Analysis using Python
# Author: Harish Mutta

import pandas as pd
import matplotlib.pyplot as plt
import os

# Create charts folder if it doesn't exist
os.makedirs("charts", exist_ok=True)

# Load Dataset
df = pd.read_excel("sales_data.xlsx")

print("FIRST 5 RECORDS")
print(df.head())

# Dataset Information
print("\nDataset Information")
print(df.info())

print("\nDataset Shape:", df.shape)

# Missing Values
print("\nMissing Values")
print(df.isnull().sum())

# Fill missing values if any
df = df.ffill()

# Remove Duplicate Records
duplicates = df.duplicated().sum()
print("\nDuplicate Records:", duplicates)

df.drop_duplicates(inplace=True)

# Total Revenue
total_revenue = df["Total_Sales"].sum()

print("\nTotal Revenue")
print(total_revenue)

# Total Orders
print("\nTotal Orders")
print(df["Order_ID"].nunique())

# Top Selling Products
top_products = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTop Selling Products")
print(top_products)

# Category Sales
category_sales = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nCategory Wise Sales")
print(category_sales)

# City Sales
city_sales = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nCity Wise Sales")
print(city_sales)

# Monthly Sales
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Month"] = df["Order_Date"].dt.month_name()

month_order = [
    "January", "February", "March",
    "April", "May", "June",
    "July", "August", "September",
    "October", "November", "December"
]

monthly_sales = (
    df.groupby("Month")["Total_Sales"]
    .sum()
    .reindex(month_order)
)

print("\nMonthly Sales")
print(monthly_sales)

# VISUALIZATIONS

# Monthly Sales Chart
plt.figure(figsize=(10, 5))
monthly_sales.plot(kind="line", marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("charts/monthly_sales.png")
plt.close()

# Category Sales
plt.figure(figsize=(8, 5))
category_sales.plot(kind="bar")
plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("charts/category_sales.png")
plt.close()

# City Sales
plt.figure(figsize=(8, 5))
city_sales.plot(kind="bar")
plt.title("City Wise Sales")
plt.xlabel("City")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("charts/city_sales.png")
plt.close()

# Top Products
plt.figure(figsize=(8, 5))
top_products.plot(kind="bar")
plt.title("Top Selling Products")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("charts/top_products.png")
plt.close()

# BUSINESS INSIGHTS

print("\nBUSINESS INSIGHTS")

print(f"Total Revenue : ₹{total_revenue:,.2f}")
print(f"Top Product : {top_products.idxmax()}")
print(f"Highest Revenue Category : {category_sales.idxmax()}")
print(f"Top Performing City : {city_sales.idxmax()}")
print(f"Best Month : {monthly_sales.idxmax()}")

print("\nCharts have been saved in the 'charts' folder.")
print("\nProject Completed Successfully!")
```
